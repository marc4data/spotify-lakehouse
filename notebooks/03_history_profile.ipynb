# History Profile

The shape and content of the Extended Streaming History itself: how the records are distributed, how much variety they hold, where they stop, and what in them is odd. Notebook 01 says what tables exist and notebook 02 answers questions about listening; this one profiles the file, so the things nobody thought to ask about have somewhere to show up.

The profile is built for one person, chosen when the report is built — `make report-03 PROFILE=<slug>` — and every figure is measured in the run. It reads the warehouse only and makes no Spotify API call.

Open the warehouse for the profile named by `SPOT_PROFILE`, then load that profile's export records once. Every section below reads them.

```python
from spotify_lakehouse import history
from spotify_lakehouse.notebook import setup

ctx = setup()
export = history.load_export(ctx)
```
